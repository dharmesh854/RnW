##### 1. Write short notes on:
- What is Data Analysis?


### -> Definition
Data Analysis is the process of inspecting, cleaning, transforming, and modeling data with the goal of discovering useful information, drawing conclusions, and supporting decision-making. It is a crucial step before building any machine learning or AI system, as it helps us understand the story hidden inside data.

### -> Purpose
The main purpose of data analysis is to:

- Understand the structure and quality of the dataset.

- Identify patterns, trends, and relationships within the data.

- Detect anomalies or outliers that may impact results.

- Support business decisions with data-driven insights.

- How to Plan a Data Science Project?


### -> Definition
A Data Science Project Plan is a structured roadmap that defines the sequence of steps to be followed when solving a problem using data. It ensures that efforts are organized, goal-driven, and measurable, preventing wasted time on unnecessary tasks.

### -> Purpose

- To provide clarity on what the project aims to achieve.

- To allocate resources (time, tools, data, and people) efficiently.

- To ensure a step-by-step workflow for solving real-world problems.

- To avoid common pitfalls such as collecting irrelevant data or overcomplicating models.

### -> Typical Workflow of a Data Science Project 🚀

##### 1. Define the Problem

- What is the business or research question?
Example: Predict customer churn for a telecom company.

##### 2. Collect Data

Identify data sources: CSV files, APIs, databases, sensors, etc.

##### 3. Data Preprocessing & Cleaning

Handle missing values, duplicates, outliers.

Transform data into an analyzable format.

##### 4. Exploratory Data Analysis (EDA)

Use descriptive statistics and visualizations to understand data.

##### 5. Feature Engineering

Create meaningful variables/features for the model.

##### 6. Model Building

Apply machine learning algorithms (e.g., regression, classification).

##### 7. Model Evaluation

Check performance using metrics like Accuracy, F1-score, RMSE.

##### 8. Deployment

Integrate the model into a real-world system (web app, dashboard, API).

##### 9. Monitoring & Maintenance

Continuously check performance and update as data changes.



- How to Frame a Machine Learning Problem?

### -> Definition
Framing a Machine Learning Problem means translating a real-world business or research question into a well-defined ML task. It involves identifying the type of problem (classification, regression, clustering, etc.), the input features, and the target variable.

### -> Purpose

- To ensure the right problem is solved with ML.

- To avoid misusing ML when traditional statistics or rule-based systems are enough.

- To clearly define inputs, outputs, and evaluation metrics before model development.

### -> Steps to Frame an ML Problem

##### 1. Understand the Business/Research Objective

What decision needs support?
Example: “Predict if a customer will churn.”

##### 2. Check if ML is Needed

Can rules or statistics solve this instead?

Example: If 100% rule-based, ML may not be necessary.

##### 3. Define Inputs (Features) and Outputs (Target Variable)

Input (X): Variables we can observe (e.g., age, salary, purchase history).

Output (y): What we want to predict (e.g., churn: yes/no).

##### 4. Select the Type of ML Task

- Classification → Categorical outcomes (spam/not spam).
- Regression → Continuous outcomes (predicting house prices).
- Clustering → Grouping unlabeled data (customer segmentation).
- Recommendation → Suggesting items (Netflix movies).

##### 5. Define Evaluation Metrics

Accuracy, Precision, Recall, F1-Score for classification.

RMSE, MAE, R² for regression.

##### 6. Set Success Criteria

Example: “Model must predict churn with at least 85% recall.”

```python
# ----- Missing Indicator -----

from sklearn.impute import MissingIndicator

num_cols = df[["age", "annual_income", "credit_score"]]

indicator = MissingIndicator(features="missing-only")

indicator_array = indicator.fit_transform(num_cols)

indicator_cols = [num_cols.columns[i] + "_missing"
                  for i in indicator.features_]

df_indicator = pd.DataFrame(
    indicator_array.astype(int),
    columns=indicator_cols,
    index=df.index
)

# --- Mean Imputation --- 
imputer = SimpleImputer(strategy="mean")

df_imputed = pd.DataFrame(
    imputer.fit_transform(num_cols),
    columns=num_cols.columns,
    index=df.index
)

# --- Merge Imputed Data + Indicator --- 
df_final = pd.concat([df_imputed, df_indicator], axis=1)

print("Final Dataset with Indicators:")
print(df_final.round(2))

print("\nMissing column indices:", indicator.features_)
print("Missing columns:", list(num_cols.columns[indicator.features_]))
```

```python
# ---------- Random Sample Imputation ----------

print("Missing Values: ", df.isnull().sum())

df_copy = df.copy()

np.random.seed(42)

if df_copy["age"].isnull().sum() > 0:
    null_count = df_copy["age"].isnull().sum()
    random_values = df["age"].dropna().sample(
        n=null_count,
        replace=True,
        random_state=42
    ).values
    df_copy.loc[df_copy["age"].isnull(), "age"] = random_values
    
    
print("age - \nOriginal Non-Null States: ")
display(df["age"].dropna().describe().round(1))

print("\nage - After Imputation Stats:")
display(df_copy["age"].dropna().describe().round(1))
```

### Part D : Outlier Handling

##### 7. Detect and treat outliers using:
- Z-Score Method
- IQR Method
- Percentile Method
- Winsorization Technique

```python
# ----- Z-Score Method -----

## Calculating Z-Score

cr_score_mean = df["credit_score"].mean()
cr_score_std = df["credit_score"].std()

print("Mean (credit_score): ",cr_score_mean)
print("Standard Deviation (credit_score): ",cr_score_std)

## Calculating Z-Score for each row of loan_amount column
df["Z-Score (credit_score)"] = (df["credit_score"] - cr_score_mean)/cr_score_std
```

```python
# ----------- Identifying Outliers -----------
threshold = 3
outliers_loan = df[df["Z-Score (credit_score)"].abs() > 3]

print(f"Outliers Detected > {threshold}:")
print(outliers_loan[['customer_id','credit_score','Z-Score (credit_score)']])
```

```python
# ----------- Removing Outliers -----------

df_clean = df[df["Z-Score (credit_score)"].abs() <= threshold].copy()
df_clean = df_clean.drop(columns=["Z-Score (credit_score)"]) 

print("Total Rows Before Cleaning: ", len(df))
print("\nRows After Cleaning Outliers (From 'credit_score'): ", len(df_clean))
print(f"Outliers removed:     {len(df) - len(df_clean)}")

print("\nMean credit_score Before: ", round(df["credit_score"].mean(), 2))
print("Mean credit_score After: ", round(df_clean["credit_score"].mean(), 2))
```

```python
# ----- IQR Method -----

plt.figure(figsize=(8,6))

plt.boxplot(df["annual_income"].dropna())

plt.show()
```

```python
## Calculating Q1 & Q3

Q1 = df["annual_income"].quantile(0.25)       # Q1
Q3 = df["annual_income"].quantile(0.75)       # Q3
IQR = Q3 - Q1

print(f"Q1 (25th Percentile): {Q1}")
print(f"Q3 (75th Percentile): {Q3}")
print(f"IQR (Q3 - Q1): {IQR}")
```

```python
## Calculating Fenses (Upper & Lower) 

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print(f"Lower Fence : {lower_fence} (Outliers below this)")
print(f"Upper Fence : {upper_fence} (Outliers above this)")
```

```python
## Identifying Outliers 

outlier_mask = (df["annual_income"] < lower_fence) | (df["annual_income"] > upper_fence)
print(f"\nOutliers in {"annual_income"} :")
print(df[outlier_mask][["customer_id", "annual_income"]])

data_mask = (df["annual_income"] > lower_fence) & (df["annual_income"] < upper_fence)
print(f"\nGenuine Data in {"annual_income"} :")
print(df[data_mask][["customer_id", "annual_income"]])
```

```python
# ----------- Removing Outliers -----------
df_clean_iqr = df[~outlier_mask].copy()

print(f"Rows Before: {len(df)} | Rows After: {len(df_clean_iqr)}")
print(f"Outliers Removed: {outlier_mask.sum()}")
print(f"\nMean 'annual_income' before: {df["annual_income"].mean():.1f}")
print(f"Mean 'annual_income' after: {df_clean_iqr["annual_income"].mean():.1f}")
```

```python
# ----- Percentile Method -----

lower_pct = 0.01
upper_pct = 0.99

lower_bound = df["loan_amount"].quantile(lower_pct)
upper_bound = df["loan_amount"].quantile(upper_pct)

print(f"\n-- Percentile Boundaries --")
print(f"1st Percentile : {lower_bound:.2f}")
print(f"99th Percentile : {upper_bound:.2f}")
print(f"Normal Range : {lower_bound} to {upper_bound}")
```

```python
## Identfying Outliers 
outlier_mask = ((df["loan_amount"] < lower_bound) | (df["loan_amount"] > upper_bound))

print("Outliers Detected:")
print(df[outlier_mask][["customer_id", "loan_amount"]])
print(f"Total Outliers: {outlier_mask.sum()}")
```

```python
## Removing Outliers 
df_cleaned_percentile = df[~outlier_mask].copy()

print(f"\n--- Results ---")
print(f"Rows before: {len(df)} | Rows after: {len(df_cleaned_percentile)}")
print(f"\nloan_amount Mean BEFORE: {df['loan_amount'].mean():.2f}")
print(f"loan_amount Mean AFTER:  {df_cleaned_percentile['loan_amount'].mean():.2f}")
print(f"\nloan_amount Min BEFORE: {df['loan_amount'].min():.2f}")
print(f"loan_amount Min AFTER:  {df_cleaned_percentile['loan_amount'].min():.2f}")
print(f"\nloan_amount Max BEFORE: {df['loan_amount'].max():.2f}")
print(f"loan_amount Max AFTER:  {df_cleaned_percentile['loan_amount'].max():.2f}")
```

```python
# ----- Winsorization Method -----

lower = df["spending_ratio"].quantile(0.05)
upper = df["spending_ratio"].quantile(0.95)

print(f"5th Percentile: {lower}")
print(f"95th Percentile: {upper}")

df["spending_ratio_winsorized"] = df["spending_ratio"].clip(lower=lower, upper=upper)

print("\nAfter Winsorization (clip):")
print(df['spending_ratio_winsorized'])
print(f"\nMean before: {df['spending_ratio'].mean():.1f}")
print(f"Max before:  {df['spending_ratio'].max():.1f}")
print(f"\nMean after: {df['spending_ratio_winsorized'].mean():.1f}")
print(f"Max after:  {df['spending_ratio_winsorized'].max():.1f}")
print("\nDataset size before: ", len(df))
print("Dataset size after: ", len(df))
```

##### 10. Encode numerical features
- Binning (discretize income into groups)
- Binarization (flag if > threshold)
- Quantile Binning
- K-Means Binning

```python
# METHOD 1: pd.cut() — Auto Equal Width (n bins)

df['income_bin'] = pd.cut(
    df['annual_income'],
    bins=5,                                  # 5 equal-width bins
    labels=['Low Income', 'Lower Middle Income', 'Middle Income', 'Upper Middle Income', 'High Income']
)

print("\nIncome Binning (3 Equal-Width Bins):")
print(df[['customer_id', 'annual_income', 'income_bin']])
```

```python
# METHOD 2: sklearn KBinsDiscretizer (Equal Width)

from sklearn.preprocessing import KBinsDiscretizer

kbd = KBinsDiscretizer(
    n_bins=4,
    encode='ordinal',     # 0,1,2,3 integers (ordinal) ya onehot
    strategy='uniform'    # equal width bins
)

df[['age_binned_sk']] = kbd.fit_transform(df[['age']])

print("\nBin edges learned:", kbd.bin_edges_)
print("\nsklearn KBinsDiscretizer (Equal Width):")
df[['customer_id', 'age', 'age_binned_sk']]
```

```python
# METHOD 1: pd.qcut() — Quantile Binning

df['credit_score_q'] = pd.qcut(
    df['credit_score'],
    q=5,                                         # 4 quartiles (Q1, Q2, Q3, Q4)
    labels=["Very Low", "Low", "Medium", "High", "Excellent"]
)

print("\nQuantile Binning (pd.qcut):")
print(df['credit_score_q'].value_counts())
```

```python
df_temp = df.dropna(subset=["credit_score"])

kbd_q = KBinsDiscretizer(n_bins=5, encode='ordinal', strategy='quantile')
df_temp[['credit_score_binned']] = kbd_q.fit_transform(df_temp[['credit_score']])

print("\nsklearn KBinsDiscretizer (quantile) bin edges:")
print(kbd_q.bin_edges_[0].round(2))

df_temp
```

```python
# ----- Binarization (flag if > threshold) -----

from sklearn.preprocessing import Binarizer

binarizer = Binarizer(threshold=700)    # Below threshold = 0, Above threshold = 1 
df_temp['credit_score_binary'] = binarizer.fit_transform(df_temp[['credit_score']])

print(df_temp[['credit_score', 'credit_score_binary']].head(10))
```

```python
df_temp["credit_score_binary"].value_counts()
```

```python
# ----- K-Means Binning -----

df_temp_kbin = df.dropna(subset=['transaction_count'])

from sklearn.preprocessing import KBinsDiscretizer

kbd_km = KBinsDiscretizer(
    n_bins=4,
    encode='ordinal',
    strategy='kmeans'
)

df_temp_kbin['transaction_kmeans_bin'] = kbd_km.fit_transform(df_temp_kbin[['transaction_count']])

print("\nK-Means Binning (KBinsDiscretizer):")
print(df_temp_kbin['transaction_kmeans_bin'].value_counts().sort_index())
print("\nK-Means bin edges:", kbd_km.bin_edges_[0].round(1))

df_temp_kbin
```
